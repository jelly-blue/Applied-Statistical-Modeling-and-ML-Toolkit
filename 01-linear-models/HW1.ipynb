```r
d <- read.csv("UsedCars.csv", header = TRUE)
summary(d)
```

```r
model1 <- lm(Price ~ Age + KM + HP + Metallic + Automatic + CC + Doors + Gears + Weight, data = d)
summary(model1)
```

```r
y <- d$Price
fit_y <- fitted(model1)
residual_m1 <- residuals(model1)

comparison_df <- data.frame(
  Original_y = y,
  Fitted_y = fit_y,
  Residuals = residual_m1
)

head(comparison_df, 10)
```

```r
# Extract coefficients table from the model summary
model_summary_table <- coef(summary(model1))

# Grab the raw estimates and standard errors
beta_hat <- model_summary_table[, "Estimate"]
se_beta  <- model_summary_table[, "Std. Error"]

# Calculate the t-statistic manually
calculated_t <- beta_hat / se_beta

# Co-list with the model's generated t-statistics
t_comparison <- data.frame(
  Calculated_t = calculated_t,
  Model_t      = model_summary_table[, "t value"]
)

print(t_comparison)
# Check if they are identical (ignoring micro floating-point issues)
all.equal(t_comparison$Calculated_t, t_comparison$Model_t)
```

```r
# 1. Get degrees of freedom
degrees_of_freedom <- df.residual(model1)
cat("Degrees of Freedom:", degrees_of_freedom, "\n")

# 2. Find the critical value cutoff (97.5th percentile for a two-tailed 95% test)
critical_value <- qt(0.975, df = degrees_of_freedom)
cat("Critical t-value:", critical_value, "\n")
```

```r
# Calculate p-values manually using the absolute value of our calculated t-stats
calculated_p <- 2 * pt(-abs(calculated_t), df = degrees_of_freedom)

# Co-list with the model's generated p-values
p_comparison <- data.frame(
  Calculated_p = calculated_p,
  Model_p      = model_summary_table[, "Pr(>|t|)"]
)

print(p_comparison)
# Check if they match
all.equal(p_comparison$Calculated_p, p_comparison$Model_p)
```

```r
# Create a summary boolean matrix to evaluate significance
significance_check <- data.frame(
  Variable = rownames(model_summary_table),
  Abs_t_Stat = abs(calculated_t),
  Critical_t = critical_value,
  p_Value = calculated_p,
  Is_Significant = calculated_p < 0.05
)

print(significance_check)
```

```r
# 1. Grab actual and fitted values from model1
y <- d$Price
y_hat <- fitted(model1)
y_bar <- mean(y)

# 2. Calculate ESS and TSS
ess <- sum((y_hat - y_bar)^2)
tss <- sum((y - y_bar)^2)

# 3. Calculate R-squared manually
calculated_r2 <- ess / tss

# 4. Extract R-squared from routine summary
routine_r2 <- summary(model1)$r.squared

cat("Manually Calculated R2:", calculated_r2, "\n")
cat("Routine Summary R2:    ", routine_r2, "\n")
all.equal(calculated_r2, routine_r2)
```

```r
# Install and load car package if you haven't already
# install.packages("car")
library(car)

# Calculate VIF for model1 estimators
vif_values <- vif(model1)
print(vif_values)
```

```r
# i. Regress Weight on all other independent variables
weight_model <- lm(Weight ~ Age + KM + HP + Metallic + Automatic + CC + Doors + Gears, data = d)
r2_weight <- summary(weight_model)$r.squared

# ii. Calculate VIF using the defining formula
vif_weight_manual <- 1 / (1 - r2_weight)

cat("Manually Calculated VIF for Weight:", vif_weight_manual, "\n")
cat("Package-Generated VIF for Weight:  ", vif_values["Weight"], "\n")
print(r2_weight)
```
