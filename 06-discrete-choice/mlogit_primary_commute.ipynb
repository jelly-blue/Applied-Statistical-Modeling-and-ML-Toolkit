```r
install.packages("mlogit")
library(mlogit)
```

```r
d <- read.csv("Commute_Mode.csv", header = TRUE)
summary(d)
```

```r
head(d)
```

```r
commute <- mlogit.data(d, shape = "long", choice = "choice", alt.var = "mode", chid.var = "id")
head(commute)
```

```r
ml.res <- mlogit(choice~cost+time, commute)
summary(ml.res)
```

```r
head(fitted(ml.res, outcome = FALSE))
```

```r
cost.avg <- tapply(commute$cost, commute$alt, mean)
cost.avg
time.avg <- tapply(commute$time, commute$alt, mean)
time.avg
xval <- data.frame(cost = cost.avg, time = time.avg)
xval
```

```r
effects(ml.res, covariate = "time", data = xval)
```
