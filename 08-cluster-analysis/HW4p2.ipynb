```r
d <- read.csv("ShoppingVisits.csv", header = TRUE)
d
plot(d, pch = 20)
text(d, labels = row.names(d), pos = 1, offset = 0.3, cex = 0.5)
```

```r
km.res.ns <- kmeans(d, 3, nstart = 20)
km.res.ns
km.res.ns$cluster
plot(d, col = km.res.ns$cluster, pch = 20, main= "K-Means no Scaling")
text(d, labels = row.names(d), col = km.res.ns$cluster, pos = 1, offset = 0.3, cex = 0.5)
```

```r
d.sc <- scale(d)
d.sc
```

```r
km.res.2 <- kmeans(d.sc, 2, nstart = 20)
km.res.2
plot(d, col = km.res.2$cluster, pch = 20, main= "K-Means With Scaling")
text(d, labels = row.names(d), col = km.res.2$cluster, pos = 1, offset = 0.3, cex = 0.5)
```

```r
km.res.3 <- kmeans(d.sc, 3, nstart = 20)
km.res.4 <- kmeans(d.sc, 4, nstart = 20)
km.res.5 <- kmeans(d.sc, 5, nstart = 20)
km.res.6 <- kmeans(d.sc, 6, nstart = 20)
```

```r
par(mfrow = c(2,2))
plot(d, col = km.res.2$cluster, pch = 20, main= "K-Means with 2 Clusters")
text(d, labels = row.names(d), col = km.res.2$cluster, pos = 1, offset = 0.3, cex = 0.5)
plot(d, col = km.res.3$cluster, pch = 20, main= "K-Means with 3 Clusters")
text(d, labels = row.names(d), col = km.res.2$cluster, pos = 1, offset = 0.3, cex = 0.5)
plot(d, col = km.res.4$cluster, pch = 20, main= "K-Means with 4 Clusters")
text(d, labels = row.names(d), col = km.res.2$cluster, pos = 1, offset = 0.3, cex = 0.5)
plot(d, col = km.res.5$cluster, pch = 20, main= "K-Means with 5 Clusters")
text(d, labels = row.names(d), col = km.res.2$cluster, pos = 1, offset = 0.3, cex = 0.5)
plot(d, col = km.res.6$cluster, pch = 20, main= "K-Means with 6 Clusters")
text(d, labels = row.names(d), col = km.res.2$cluster, pos = 1, offset = 0.3, cex = 0.5)
```

```r
ss1 <- km.res.2$totss
ss2 <- km.res.2$tot.withinss
ss3 <- km.res.3$tot.withinss
ss4 <- km.res.4$tot.withinss
ss5 <- km.res.5$tot.withinss
ss6 <- km.res.6$tot.withinss
ss.vec <- c(ss1,ss2,ss3,ss4,ss5,ss6)
par(mfrow = c(1,1))
plot(ss.vec, type = "b", xlab = "Number od Clusters", ylab = "Total within-Cluster SS")
```

```r
plot(d, col = km.res.3$cluster, pch = 20, main= "K-Means with 3 Clusters")
text(d, labels = row.names(d), col = km.res.2$cluster, pos = 1, offset = 0.3, cex = 0.5)
```

```r
hc.res.ave <- hclust(dist(d.sc), method = "average")
plot(hc.res.ave, cex = 0.5, main="Dendrogram (Average Linkage)", xlab = "", ylab = "", sub = "")
abline(h=1.75, col="red", lty = 2)
```

```r
hc.res.cen <- hclust(dist(d.sc), method = "centroid")
plot(hc.res.cen, cex = 0.5, main="Dendrogram (Centroid Linkage)", xlab = "", ylab = "", sub = "")
abline(h=.9, col="red", lty = 2)
```

```r
hc.res.single <- hclust(dist(d.sc), method = "single")
plot(hc.res.single, cex = 0.5, main="Dendrogram (Single Linkage)", xlab = "", ylab = "", sub = "")
abline(h=.51, col="red", lty = 2)
```

```r
hc.res.comp <- hclust(dist(d.sc), method = "complete")
plot(hc.res.comp, cex = 0.5, main="Dendrogram (Complete Linkage)", xlab = "", ylab = "", sub = "")
abline(h=3.2, col="red", lty = 2)
```

```r
cutree(hc.res.ave, k=3)

par(mfrow = c(1,2))
plot(d, col=km.res.3$cluster, pch=20, main=paste("K-means with 3 clusters"))
text(d, labels = row.names(d), col=km.res.3$cluster, pos=1, offset=0.3, cex = 0.5)
plot(d, col = cutree(hc.res.ave, k=3), pch=20, main="Hiearchial Clustering (Average Linkage)")
text(d, labels = row.names(d), col=cutree(hc.res.ave, k=4), pos=1, offset=0.3, cex = 0.5)
```

```r
par(mfrow = c(1,2))
plot(d, col = cutree(hc.res.ave, k=3), pch=20, main="Hiearchial Clustering (Average Linkage)")
text(d, labels = row.names(d), col=cutree(hc.res.ave, k=4), pos=1, offset=0.3, cex = 0.5)
plot(d, col = cutree(hc.res.comp, k=3), pch=20, main="Hiearchial Clustering (Complete Linkage)")
text(d, labels = row.names(d), col=cutree(hc.res.comp, k=4), pos=1, offset=0.3, cex = 0.5)
plot(d, col = cutree(hc.res.single, k=3), pch=20, main="Hiearchial Clustering (Single Linkage)")
text(d, labels = row.names(d), col=cutree(hc.res.single, k=4), pos=1, offset=0.3, cex = 0.5)
plot(d, col = cutree(hc.res.cen, k=3), pch=20, main="Hiearchial Clustering (Centroid Linkage)")
text(d, labels = row.names(d), col=cutree(hc.res.cen, k=4), pos=1, offset=0.3, cex = 0.5)
```
